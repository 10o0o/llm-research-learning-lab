```python
def observe_shape(name: str, value: np.ndarray) -> dict[str, object]:
    """Return a compact name and shape card."""
    if not isinstance(value, np.ndarray):
        raise TypeError("value must be a NumPy array")

    # TODO: 배열의 shape를 일반 tuple로 바꾸세요.
    shape = NotImplemented
    return {"name": name, "shape": shape}
```

```python
# 예제 입력으로 동작을 살펴보세요
print(observe_shape("sample", np.zeros((2, 3))))
```

```python
# 구현을 마친 뒤 이 셀을 실행하세요
def check_e01() -> None:
    card = observe_shape("sample", np.zeros((2, 3)))
    np.testing.assert_equal(card["name"], "sample")
    np.testing.assert_equal(card["shape"], (2, 3))

    scalar = observe_shape("scalar", np.array(1.0))
    np.testing.assert_equal(scalar["shape"], ())

    try:
        observe_shape("bad", [1, 2])
    except TypeError:
        pass
    else:
        raise AssertionError("NumPy 배열이 아니면 TypeError여야 합니다")


check_e01()
```
